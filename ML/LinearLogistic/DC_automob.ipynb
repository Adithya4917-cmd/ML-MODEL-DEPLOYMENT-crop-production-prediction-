```python
#price — 0 means the car is free, which is misleading. Better to use mean or median
#that meaning there are some string in the prices so covertion needed
#then we can find the mean or median
#that means we need to remove $ signs
df['price'].unique()
```

```python
df['price'].value_counts()
```

```python
#to remove the signs $

df['price'] = df['price'].str.replace('$', '', regex=False)
```

```python
df['price'] .unique()
```

```python
df['price'].value_counts() 
```

```python
df['price'] = pd.to_numeric(df['price'], errors='coerce')
#  errors='coerce' this means that if '15885' → converts fine → 15885
#                                     '$24970' → can't convert (has $) → becomes NaN
```

```python
df['price'].value_counts()
```

```python
#now to remove the stupid values like -500, 10000000 and 0
        # the reason we are doing the 0 to nan is because they affect our median's value
import numpy as np
condition = (df['price']>=0) & (df['price']>99999)
df.loc[condition, 'price'] = np.nan
```

```python
print(df['price'].isna().sum())
#shows number of nan values
```

```python
print(df['price'])
```

```python
df['price'].describe()
#with this you can know what to apply to ur nan values --mean /--median /--mode
```

```python
#median is correct
df['price']=df['price'].fillna(df['price'].median())
```

```python
print(df['price'].isna().sum())
#shows number of nan values
# 0 meaning nan is filled with median
```
