```python
df["Glucose"].value_counts()
```

```python
df["Glucose"] = df["Glucose"].str.replace("#","",regex=False)
df["Glucose"] = df["Glucose"].str.replace("O","",regex=False)
df["Glucose"] = df["Glucose"].str.replace("l","",regex=False)
df["Glucose"] = df["Glucose"].str.replace("?","",regex=False)
```

```python
df["Glucose"].unique()
```

```python
df["Glucose"] = pd.to_numeric(df["Glucose"] , errors='coerce').astype('Int64')
```

```python
df["Glucose"].unique()
```

```python
df["Glucose"] = df["Glucose"].fillna(df["Glucose"].median())
```

```python
df["Glucose"].isna().sum()
```
